# Funciones
### Una funcion agrupa codigo reutilizable. Se define con def

```python
def clasificar_edad (edad):
    """Recibe una edad y regresa la categoria correspondiente"""
    if edad <12:
        return "Niño"
    elif edad <18:
        return "Adolescente"
    elif edad <65:
        return "Adulto"
    else:
        return "Adulto mayor"

# probamos la funcion con distintos valores
for edad in [5, 15, 40, 80]:
    print (edad, "->", clasificar_edad(edad))
```

```python
def agregar_pasajeros():
    lista_nueva = []
    maximo_pasajeros = 4
    
    for i in range(maximo_pasajeros):
        espacios_restantes = maximo_pasajeros - len(lista_nueva)
        if espacios_restantes > 0:
            print("quedan ", espacios_restantes, "espacios disponibles")
            
            #datos
            nombre = input("Ingresa el nombre del pasajero: ")
            edad = int(input("Ingresa la edad: "))
            clase = int(input("Ingresa la clase (1, 2 o 3): "))
            tarifa = float(input("Ingresa la tarifa: "))
            
            #Diccionario
            nuevo_pasajero = {
                "nombre": nombre,
                "edad": edad,
                "clase": clase,
                "Tarifa": tarifa
            }
            #agregar 
            lista_nueva.append(nuevo_pasajero)
            
            print("Se agrego pasajero: ", nuevo_pasajero["nombre"])
            
            #preguntar si agrega mas
            if maximo_pasajeros - len(lista_nueva) > 0:
                continuar = input("¿Deseas agregar otro pasajero? 1 para si, 0 para no: ")
                if continuar != "1":
                    break
                    
    #Mostrar
    if len(lista_nueva) == maximo_pasajeros:
        print("se lleno la lista")
        
    return lista_nueva

# Ejecutar el registro
pasajeros_registrados = agregar_pasajeros()

print("ESTADO FINAL")
print("Total registrados:", len(pasajeros_registrados))
print("Pasajeros en la lista:", len(pasajeros_registrados))

print("Lista de pasajeros registrados:")
for p in pasajeros_registrados:
    print("Nombre:", p["nombre"], "- Edad:", p["edad"], "- Clase:", p["clase"], "- Tarifa:", p["Tarifa"])
```

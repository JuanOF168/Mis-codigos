# Limpieza basica de datos

tratar los valores faltantes: Rellenar las columnas que tienen datos faltantes coon la media de los demas

Dropear: cabin, embarked

Convertir los tipos: Asegurar que la columna Pclass sea categorico

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
 #Rellenar columnas de Age
```

```python
df.head(10)
```

```python
 df.dropna(subset = ['Embarked'], inplace=True) #Eliminar filas con valores vacios
```

```python
df.drop('Cabin', axis=1, inplace=True) #Eliminar la columna cabina
```

```python
df['Pclass'] = df['Pclass'].astype('category') #Convertit
```

```python
df.info()
```

# Analisis univariado: Distribuciones (graficas)

```python
#configuracion estetica
import seaborn as sns
sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10,6)
```

```python
#Variable objetivo
sns.countplot(x="Survived", data=df)
plt.title("Distribucion de sobrevivientes")
plt.show()
```

```python
#variables categoricas
cat_vars = ["Sex", "Pclass", "Embarked"]
```

```python
for col in cat_vars:
    sns.countplot(x=col, data=df)
    plt.title("Distribucion de {col}")
    plt.show()
```

```python
#variables numericas
num_vars=["Age", "Fare"]
```

```python
for col in num_vars:
    sns.histplot(df[col].dropna(), kde=True, bins=30)
    plt.title("Distribucion de {col}")
    plt.show()
```

## Anlisis de dos variables (Bivariable)

```python
sns.barplot(x="Sex", y="Survived", data=df)
plt.title("Sobrevivientes segun sexo")
plt.show()
```

```python
#supervivencia por clase
sns.barplot(x="Pclass", y="Survived", data=df)
plt.title("Sobrevivientes segun clase")
plt.show()
```

```python
#superviviencia por edad
sns.countplot(x="precio_gbp", data=df)
plt.title("Distribución de precio_gbp")
plt.show()
```

```python
#correlaciones
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Mapa de correlaciones")
plt.show()
```
